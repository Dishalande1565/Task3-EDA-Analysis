```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Map species numbers to names
df['species_name'] = df['species'].map({
    0: 'setosa', 
    1: 'versicolor', 
    2: 'virginica'
})

print("Shape:", df.shape)
print(df.head())
```

```python
# Histogram for all features
df.drop(columns=['species','species_name']).hist(
    figsize=(10,6), 
    bins=20, 
    color='steelblue', 
    edgecolor='black'
)
plt.suptitle('Histogram - Distribution of All Features')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,5))
for i, col in enumerate(iris.feature_names):
    plt.subplot(1, 4, i+1)
    sns.boxplot(y=df[col], color='lightblue')
    plt.title(col[:10])
plt.suptitle('Box Plots - Outlier Detection')
plt.tight_layout()
plt.show()
```

```python
# Scatter plot - Sepal Length vs Petal Length
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='sepal length (cm)', 
    y='petal length (cm)',
    hue='species_name', 
    data=df,
    palette='Set1'
)
plt.title('Sepal Length vs Petal Length by Species')
plt.show()
```

```python
# Average feature values per species
plt.figure(figsize=(10,5))
df.groupby('species_name')[iris.feature_names].mean().plot(
    kind='bar', 
    figsize=(10,5),
    colormap='Set2',
    edgecolor='black'
)
plt.title('Average Feature Values per Species')
plt.xticks(rotation=0)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(
    df.drop(columns=['species','species_name']).corr(),
    annot=True, 
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
sns.pairplot(
    df.drop(columns=['species']), 
    hue='species_name',
    palette='Set1',
    diag_kind='kde'
)
plt.suptitle('Pairplot - All Features vs All Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='species_name', data=df, palette='Set2')
plt.title('Species Count Distribution')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

## 🎯 Interview Questions & Answers

### Q1. Univariate vs Bivariate vs Multivariate Analysis?
- **Univariate:** Analyse ONE variable at a time
  - Example: Histogram of Age, Boxplot of Fare
- **Bivariate:** Relationship between TWO variables
  - Example: Scatter plot of Height vs Weight
- **Multivariate:** Multiple variables together
  - Example: Pairplot of all features by species

### Q2. How do you interpret a Correlation Heatmap?
- Values range from -1 to +1
- **+1:** Perfect positive correlation (both increase)
- **0:** No correlation
- **-1:** Perfect negative correlation (one increases, other decreases)
- Dark red = strong positive, Dark blue = strong negative

### Q3. What is the purpose of a Box Plot?
- Shows distribution of data in 5 numbers:
  - Minimum, Q1 (25%), Median, Q3 (75%), Maximum
- Points outside whiskers = Outliers
- Box size shows spread of middle 50% data

### Q4. How do you identify outliers visually?
- **Box Plot:** Points beyond whiskers are outliers
- **Scatter Plot:** Points far from cluster are outliers
- **Histogram:** Very long tails show outliers

### Q5. Why is EDA the most important step before modeling?
- Understand data structure and patterns
- Find and fix missing values
- Detect outliers that affect model
- Find important features for prediction
- Garbage in = Garbage out in ML!
